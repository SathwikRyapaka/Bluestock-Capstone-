# Bluestock Fintech — Quantitative Performance & Benchmark Attribution

This notebook implements the full quantitative performance evaluation across all 40 mutual fund schemes:
1. **Daily Returns** (`nav_t / nav_{t-1} - 1`) & distribution validation
2. **1-Year, 3-Year, and Full-Period (~4.4Y/5Y) CAGR** comparison
3. **Sharpe Ratio** ($R_f = 6.5\%$) & **Sortino Ratio** (downside deviation only)

5. **Maximum Drawdown** & worst peak-to-trough date ranges
6. **Composite Fund Scorecard (0–100)**
7. **3-Year Benchmark Comparison & Tracking Error** vs NIFTY 50 & NIFTY 100

```
import pandas as pd
from IPython.display import Image, display
from generate_performance_analytics import (
    load_nav_and_benchmarks, compute_daily_returns,
    compute_performance_metrics, plot_benchmark_comparison
)

df_nav, df_bench, df_perf = load_nav_and_benchmarks()
df_nav = compute_daily_returns(df_nav)
df_scorecard, df_alpha_beta = compute_performance_metrics(df_nav, df_bench, df_perf)
plot_benchmark_comparison(df_nav, df_bench, df_scorecard)
```
